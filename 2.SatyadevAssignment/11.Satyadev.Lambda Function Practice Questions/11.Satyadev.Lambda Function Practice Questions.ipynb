```python
def sqrNum(x):  # parameters
    return x**2

sqrNum(10)
```

```python
nums = [1, 2, 3, 4]

addFive = lambda x: x + 5

result = list(map(addFive, nums))

print(result)
```

```python
nums = [5, 8, 11, 14, 17, 20]

evenNum = lambda x: x % 2 == 0

result = list(filter(evenNum, nums))

print(result)
```

```python
from functools import reduce

nums = [2, 3, 4, 5]

multiply = lambda x, y: x * y

result = reduce(multiply, nums)

print(result)
```

```python
data = [(1, 3), (2, 1), (4, 2)]

result = sorted(data, key=lambda x: x[1])

print(result)
```

```python
checkA = lambda x: x.startswith('A')

print(checkA("Apple"))
print(checkA("Banana"))
```

```python
temps = [0, 20, 37, 100]

convert = lambda x: (x * 9/5) + 32

result = list(map(convert, temps))

print(result)
```

```python
words = ['pen', 'notebook', 'apple', 'go']

checkLength = lambda x: len(x) > 3

result = list(filter(checkLength, words))

print(result)
```

```python
average = lambda x, y: (x + y) / 2

print(average(10, 20))
```

```python
words = ['python', 'lambda', 'function']

capitalizeWord = lambda x: x.capitalize()

result = list(map(capitalizeWord, words))

print(result)
```

```python
result = [x**2 for x in range(1, 11)]

print(result)
```

```python
result = [x for x in range(1, 51) if x % 2 == 0]

print(result)
```

```python
words = ['cat', 'dog', 'lion']

result = [x.upper() for x in words]

print(result)
```

```python
result = [x for x in range(1, 101) if x % 3 == 0 and x % 5 == 0]

print(result)
```

```python
text = "Comprehension"

result = [x for x in text if x.lower() in 'aeiou']

print(result)
```

```python
matrix = [[1,2,3], [4,5,6], [7,8,9]]

result = [x for row in matrix for x in row]

print(result)
```

```python
words = ['tree', 'book', 'python', 'sky']

result = [x for x in words if 'e' in x]

print(result)
```

```python
result = [x for x in range(100, 1000) if str(x) == str(x)[::-1]]

print(result)
```

```python
nums = [1, 2, 3, 4, 5, 6]

result = ['ODD' if x % 2 != 0 else x for x in nums]

print(result)
```

```python
result = [[1 if i == j else 0 for j in range(3)] for i in range(3)]

print(result)
```
