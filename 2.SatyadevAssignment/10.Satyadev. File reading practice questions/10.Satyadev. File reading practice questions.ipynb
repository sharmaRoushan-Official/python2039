```python
file = open("data.txt", "r")

data = file.read()

count = 0

for x in data:
    if x.lower() in "aeiou":
        count = count + 1

print("Total vowels:", count)

file.close()
```

```python
file = open("data.txt", "r")

data = file.read()

upper = 0
lower = 0

for x in data:

    if x.isupper():
        upper = upper + 1

    if x.islower():
        lower = lower + 1

print("Uppercase letters:", upper)
print("Lowercase letters:", lower)

file.close()
```

```python
file = open("data.txt", "r")

data = file.read()

words = data.split()

longest = ""

for word in words:
    if len(word) > len(longest):
        longest = word

print("Longest word:", longest)

file.close()
```
