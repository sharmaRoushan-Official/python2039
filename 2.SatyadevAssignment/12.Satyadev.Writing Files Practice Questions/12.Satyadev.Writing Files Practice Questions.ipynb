```python
file = open("even.txt", "w")

for x in range(1, 51):
    if x % 2 == 0:
        file.write(str(x) + "\n")

file.close()
```

```python
file = open("odd.txt", "w")

for x in range(1, 51):
    if x % 2 != 0:
        file.write(str(x) + "\n")

file.close()
```

```python
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]

file = open("fruits.txt", "w")

for fruit in fruits:
    file.write(fruit + "\n")

file.close()
```
