1.Student Registration Validation

```python
class ValidationError(Exception):
    pass

name = input("Enter Name: ")
age = int(input("Enter Age: "))
email = input("Enter Email: ")

errors = []

if not name.isalpha() or len(name) < 3:
    errors.append("Name Invalid")

if age < 18 or age > 30:
    errors.append("Age Invalid")

if "@" not in email or ".com" not in email:
    errors.append("Email Invalid")

try:
    if errors:
        raise ValidationError(", ".join(errors))
    else:
        print("Registration Successful")
except ValidationError as e:
    print("Custom Exception:", e)
```

️2.Employee Registration

```python
class EmployeeError(Exception):
    pass

emp_id = input("Enter Employee ID: ")
salary = float(input("Enter Salary: "))
experience = float(input("Enter Experience (Years): "))

errors = []

if not emp_id.startswith("EMP"):
    errors.append("Invalid Employee ID")

if salary <= 20000:
    errors.append("Salary Invalid")

if experience < 1:
    errors.append("Experience Invalid")

try:
    if errors:
        raise EmployeeError(", ".join(errors))
    else:
        print("Employee Registered")
except EmployeeError as e:
    print(e)
```

3.ATM Withdrawal

```python
class ATMError(Exception):
    pass

pin = input("Enter PIN: ")
balance = float(input("Enter Balance: "))
amount = float(input("Enter Withdrawal Amount: "))

errors = []

if not (pin.isdigit() and len(pin) == 4):
    errors.append("Invalid PIN")

if amount > balance:
    errors.append("Insufficient Balance")

if amount < 100:
    errors.append("Minimum Withdrawal is ₹100")

try:
    if errors:
        raise ATMError(", ".join(errors))
    else:
        print("Withdrawal Successful")
except ATMError as e:
    print(e)
```

4.Online shopping order

```python
class OrderError(Exception):
    pass

quantity = int(input("Enter Quantity: "))
mobile = input("Enter Mobile Number: ")
address = input("Enter Address: ")

errors = []

if quantity <= 0:
    errors.append("Invalid Quantity")

if not (mobile.isdigit() and len(mobile) == 10):
    errors.append("Invalid Mobile Number")

if address.strip() == "":
    errors.append("Address Cannot Be Empty")

try:
    if errors:
        raise OrderError(", ".join(errors))
    else:
        print("Order Placed Successfully")
except OrderError as e:
    print(e)
```

5.College Admission

```python
class AdmissionError(Exception):
    pass

marks = float(input("Enter Marks: "))
age = int(input("Enter Age: "))
course = input("Enter Course Name: ")

errors = []

if marks < 0 or marks > 100:
    errors.append("Invalid Marks")

if age < 17:
    errors.append("Age Invalid")

if course.strip() == "":
    errors.append("Course Name Cannot Be Empty")

try:
    if errors:
        raise AdmissionError(", ".join(errors))
    else:
        print("Admission Successful")
except AdmissionError as e:
    print(e)
```

6. Password Validation

```python
class PasswordError(Exception):
    pass

password = input("Enter Password: ")

errors = []

if len(password) < 8:
    errors.append("Minimum 8 Characters")

if not any(ch.isupper() for ch in password):
    errors.append("Uppercase Missing")

if not any(ch.islower() for ch in password):
    errors.append("Lowercase Missing")

if not any(ch.isdigit() for ch in password):
    errors.append("Digit Missing")

special = "!@#$%^&*()_+-=<>?/"

if not any(ch in special for ch in password):
    errors.append("Special Character Missing")

try:
    if errors:
        raise PasswordError(", ".join(errors))
    else:
        print("Password Accepted")
except PasswordError as e:
    print(e)
```

7️.Bank Account Opening

```python
class BankError(Exception):
    pass

name = input("Enter Name: ")
aadhaar = input("Enter Aadhaar Number: ")
pan = input("Enter PAN Number: ")

errors = []

if not name.isalpha():
    errors.append("Invalid Name")

if not (aadhaar.isdigit() and len(aadhaar) == 12):
    errors.append("Invalid Aadhaar Number")

if len(pan) != 10:
    errors.append("Invalid PAN Number")

try:
    if errors:
        raise BankError(", ".join(errors))
    else:
        print("Account Opened Successfully")
except BankError as e:
    print(e)
```

8.Train Ticket Booking

```python
class TicketError(Exception):
    pass

name = input("Enter Passenger Name: ")
age = int(input("Enter Age: "))
tickets = int(input("Enter Number of Tickets: "))

errors = []

if name.strip() == "":
    errors.append("Name Cannot Be Empty")

if age <= 5:
    errors.append("Age Invalid")

if tickets < 1 or tickets > 6:
    errors.append("Invalid Number of Tickets")

try:
    if errors:
        raise TicketError(", ".join(errors))
    else:
        print("Ticket Booked Successfully")
except TicketError as e:
    print(e)
```

9. Library Membership

```python
class LibraryError(Exception):
    pass

member_id = input("Enter Membership ID: ")
fine = float(input("Enter Fine Amount: "))
phone = input("Enter Phone Number: ")

errors = []

if not member_id.startswith("LIB"):
    errors.append("Invalid Membership ID")

if fine < 0:
    errors.append("Fine Cannot Be Negative")

if not (phone.isdigit() and len(phone) == 10):
    errors.append("Invalid Phone Number")

try:
    if errors:
        raise LibraryError(", ".join(errors))
    else:
        print("Membership Approved")
except LibraryError as e:
    print(e)
```
