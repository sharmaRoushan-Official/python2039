```python
# Q1. Write a program to count total number of vowels and consonent in the string total length of characters
# ? 

data = "The Quick Brown Fox Jump Over The Lazy Dog"

totalLength = 0

vowelChar = []

consonentChar=[]

for ch in data:
    if ch == " ":
        continue
    totalLength+=1
    # print(ch)
    if ch in "AEIOUaeiou":
        vowelChar.append(ch)
        # print(ch)
    else:
        consonentChar.append(ch)
print("Total Length of Chracters:",totalLength)
print("Vowel Characters:",vowelChar)
print("Consonent Characters:",consonentChar)
```
