```python
def getWeather():
    api_keys = "f3d23ab84146eaf779dbe1fcb07bc122"
    city = input("Enter Country/City/State: ")
    url = f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid={api_keys}"
    print(url)
    response = requests.get(url)
    print(response)

    if response.status_code ==200:
        data  = response.json()
        print(data)
        
        
        print(f"=================Details {city}=================")
        print("=================Details Temperature=================")
        temp=data["main"]['temp']
        # 0K − 273.15 = -273.1°C
        celcuisTemp = temp - 273.15
        print("Temperature:",celcuisTemp)
        
        feelsLike = data['main']["feels_like"]
        # print(feelsLike)
        feelsLikeCelcius = feelsLike-273.15
        print("Feels Like:",feelsLikeCelcius)
        

        # print(temp)
getWeather()
```
