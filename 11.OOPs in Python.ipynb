# ➤ 1. Inheritance 

#### 1. Single/Simple Inheritance
Meaning:

* In `Single Inheritance`, a child inherits from one one parent class.

📌 Structure
`Parent ➽ Child`

#### Why it is used:
 * Code reusability
 * Child class can use parent class properties and method.
 * Easy to userdstand and maintain

#### Real-Life Example

* Parent - Vehicle
* Child - Car

`A Var is a Vehicle`

````python
class Vehicle:  # parent class
    def start(self):
        print("Vehicle Started.")


class Car(Vehicle):  # Child / Derived Class
    def drive(self):
        print("Car is Driving.")


# v = Vehicle()
# v.start()

# c = Car()
# c.drive()


c = Car()
c.start()  # start is the method of parent class
c.drive()  # drive is the method of child class
````

Note:
Single Inheritance is most basic and safest inheritance. It avoids Complixity and ambiguity.

[Ambiguity in Object-Oriented Programming (OOP) occurs when a compiler or program cannot decide which function, variable, or path to choose]    

### ★ 2. Multiple Inheritance
Meaning

In `Multiple Inheritance`,a child class inherits from more than one parent class.

📌 Structure

```text
    Parent1           Parent2
        \              /
         \            /
          \          /
             Child
```

### Why It is used?
 * When one class needs feature of multiple classes.
 * Helps to combine functionlity.

### Real-Life Example
 * Parent1: Camera
 * Parent2: Phone
 * Child: SmartPhone

`SmartPhone is both a Phone and a Camera`

Python Example

````python
class Camera: # Parent 1
    def click_photo(self):
        print("Photo Clicked......")


class Phone:  # Parent 2
    def calling(self):
        print("Calling......")


class SmartPhone(Camera,Phone): # Derived class : Parent1 + Parent2
    def internet(self):
        print("Google Connected")


obj = SmartPhone()
obj.click_photo()
obj.calling()
obj.internet()
````

Note:

Python support multiple inheritance, you can check order of method called Method Resoluction Order(MRO) to avoid confustion.



````python
print(SmartPhone.mro())
````

### ★ 3. Multilevel Inheritance

Meaning

In `Multilevel Inheritance`, inheritance happens in levels.


📌 Structure

`Grandparent ➽ Parent ➽ Child`

#### Why it is used?
 * To represent hierarchy in real world.
 * Step-by-step Inheritance of feature.

#### Real-Life Example

 * Grandparent: Animal
 * Parent: Mammal
 * Child: Dog

`Dog is a Mammal, Mammal is an Animal`

````python
class Animal:
    def eat(self):
        print("Eating......")

class Mammal(Animal):
    def walk(self):
        print("Walking")

class Dog(Mammal):
    def bark(self):
        print("Barking....")


d = Dog()
d.eat() # eat is the method of animal
d.walk() # walk is the method of mammal
d.bark() # bark is the mehtod of dog.
````

````python
print(Dog.mro())
````

### ★ 4.Hierarchical Inheritance

Meaning 

In `Hierarchical Inheritance,`, Multiple classes inherits from a single parent.


📌 Structure

```text

       Parent1
      /   |   \
     /    |    \
    /     |     \
 child1  child2  child3
 ```

### Why it is used?
* One parent has common functionality.
* Different children use and extend it differently.

#### Real-Life Example:
* Parent: Employee
* Children:
     * Manager
     * Developer
     * Tester
* All are employees but have different roles.

 Python Example

````python
class Employee:
    def work(self):
        print("Employee is working..........")

class Manager(Employee):
    def manage(self):
        print("Manager is Managing.......")

class Developer(Employee):
    def code(self):
        print("Developer is coding..........")
        
class Tester(Employee):
    def test(self):
        print("Tester is testing............")

m = Manager()
m.work()  # work() is the method of parent class
m.manage() # manage() is the method of Manager.

d = Developer()
d.work() # work() is the method of parent class.
d.code() # code() is the method of Developer class

t = Tester()
t.work() # work() is the method of parent class.
t.test() # test() is the method of Tester() class
````

````python
print(Manager.mro())
````

````python
print(Developer.mro())
````

````python
print(Tester.mro())
````

#### ★ 5. Hybrid Inheritance

meaning

Hybrid Inheritance is combination of two or more inheritance type.

It may Include:
* Multiple+Hierarchical
* Hierarchical+Multilevel

```text
          A
        /   \
       B     C
        \   /
          D 
```
Here,
* B and C inherits from A(Hierarchcal Inheritance)
* D inherits from B and C(Multiple Inheritance)

### Why it is used?
* When a system requires complex Inheritance.
* Common in large framework and enterprises 

Python Example

````python
class A:
    def showA(self):
        print("Class A")
class B(A):
    def showB(self):
        print("Class B")
class C(A):
    def showC(self):
        print("Class C")
class D(B,C):
    def showD(self):
        print("Class D")


obj = D()
obj.showA()
obj.showB()
obj.showC()
obj.showD()
````

Note:

Hybrid Inheritance is powerful but can cause confusion

 * Method Conflict
 * ambiguity

````python
print(D.mro())
````

#### 1. Method Overloading(Conceptual in Python) (Same Method or Function but different parameter)

Python does not support traditional overloading, but we can achiever using default argument.

````python
class Calculator:
    def add(self, a,b=0,c=0):  #  3 parameters 
        return a+b+c # 5+6+0


obj1 = Calculator()
print(obj1.add(5,))  # 1 parameter
print(obj1.add(5,6,)) # 2 parameter
print(obj1.add(1,2,3))  # 3 parameter
````

### Keypoints:
* Same method name add
* Different numbers of arugment
* Behavior changes automatically according to value.

### 2. Method Overriding(Same Method and same parameter: Same Signature) (Runtime Polymorphism)

Child class provides its own implementation of parent method 

````python
class Employee:
    def role(self):
        return "I am an Employee."

class Developer(Employee):
    def role(self):
        return "I am Developer."

class TeamLead(Developer):
    def role(self):
        return "I am a Team Leader"

# obj = Developer()
# obj.role()

obj1 = TeamLead()
obj1.role()
````

### Keypoints:

* Same Method Name role()
* Different Output
* Based on object type
* Runtime Polymorphism

````python
e = Employee()
print(e.role())

d = Developer()
print(d.role())

t = TeamLead()
print(t.role())
````

#### Encapsulation using Getter and setter method 

````python
class Student:
    def __init__(self):
        self.__marks = 0
    # setter method 
    def set_marks(self,marks1):  # private variable update 
        if marks1>0 and marks1<100:
            self.__marks = marks1
            print(f"Your marks is updated:{marks1}")
        else:
            print(f"Your marks {marks1} is greater than 100")

    def get_marks(self):  # only display current marks
        return f"Your Current Marks:{self.__marks}"
````

##  Encapsulation using @property Method 

````python
class Student:
    def __init__(self):
        self.__marks = 0

    @property
    def marks(self):  # only display current marks
        return f"Your Current Marks:{self.__marks}"
        
    # setter method 
    @marks.setter
    def marks(self,marks1):  # private variable update 
        if marks1>0 and marks1<100:
            self.__marks = marks1
            print(f"Your marks is updated:{marks1}")
        else:
            print(f"Your marks {marks1} is greater than 100")
    
````

````python
obj = Student()
obj.marks = 99  # setter
````

````python
obj.marks  # getter method
````

````python
obj1 = Student()
obj1.marks
````

````python
obj1.marks = 85
````

✅️ With Abstraction

  * This is Data Abstraction
 
|Feature|Encapsultion|Data Abstraction|

|Achieved by| Access Modifiers| from abc,abstractmethod/ interfaces|


>Encapsulation =  How is stored

>Abstraction = What user can do

How Abstraction is Achieved in Python

Python Provides:
1. Abstract base classes (abc) Module
2. Abstractmethod [Built-in Decorator]

1. Abstract Base Class (Core Concept):
   * Can't be instantiated.
   * Contain abstract method(built-in)
   * Acts as bluepint (interface)
2. Abstractmethod [Built-in Decraotr]
   * Has declaration only
   * No implementation
   * Must be implemented by child class

Syntax for import Module

`from abc import ABC,abstractmethod`

````python
from abc import ABC,abstractmethod


class Vehicle(ABC):

    @abstractmethod    
    def start(self):
        pass

    @abstractmethod
    def abcd(self):
        pass


# obj = Vehicle()
# obj.start()
````

````python
class Car(Vehicle):
    def automaticCar(self):
        print("My Car is Automated.")


    def start(self):
        print("Car Started.")

    def abcd(self):
        pass

obj = Car()
obj.automaticCar()
obj.start()
````

### Key Points

* Vehicle defines what should be done.
* Car defines how it is done.
* User only calls autometicCar()

### Parent constractor using `super()`

````python
class Person:
    def __init__(self,name):
        self.name = name
obj = Person("Swadhin")
obj.name
````

````python
class Employee(Person):
    def __init__(self,name,age):
        # self.name = name
        self.age = age
        super().__init__(name)
obj = Employee("Swadhin",88)
print(obj.name)
print(obj.age)
````

# Python Dunder (Magic) Methods

## Object Creation & Initialization
| Method | Purpose |
|---|---|
| `__new__(cls, ...)` | Creates a new instance (called before `__init__`) |
| `__init__(self, ...)` | Initializes a new instance |
| `__del__(self)` | Called when an object is about to be destroyed |

## Representation
| Method | Purpose |
|---|---|
| `__repr__(self)` | Official string representation (for developers) |
| `__str__(self)` | Informal string representation (for users, via `print()`) |
| `__format__(self, format_spec)` | Custom behavior for `format()` / f-strings |
| `__bytes__(self)` | Byte-string representation |

## Comparison
| Method | Purpose |
|---|---|
| `__eq__(self, other)` | `==` |

### Quick example — using several together

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"'{self.title}' ({self.pages} pages)"

    def __repr__(self):
        return f"Book({self.title!r}, {self.pages})"

    def __len__(self):
        return self.pages

    def __eq__(self, other):
        return self.pages == other.pages

    def __lt__(self, other):
        return self.pages < other.pages

b1 = Book("Python 101", 250)
b2 = Book("Deep Dive", 400)

print(b1)          # 'Python 101' (250 pages)  -> __str__
print(len(b1))      # 250                       -> __len__
print(b1 < b2)       # True                      -> __lt__
print([b1])          # [Book('Python 101', 250)] -> __repr__
```

The top 5 you'll use almost daily as a beginner/intermediate coder: `__init__`, `__str__`, `__repr__`, `__len__`, and `__eq__`.
