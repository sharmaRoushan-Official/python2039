# ➤ 1. Inheritance 

#### 1. Single/Simple Inheritance
Meaning:

* In `Single Inheritance`, a child inherits from one one parent class.

📌 Structure
`Parent ➽ Child`

#### Why it is used:
 * Code reusability
 * Child class can use parent class properties and method.
 * Easy to userdstand and maintain

#### Real-Life Example

* Parent - Vehicle
* Child - Car

`A Var is a Vehicle`

````python
class Vehicle:  # parent class
    def start(self):
        print("Vehicle Started.")


class Car(Vehicle):  # Child / Derived Class
    def drive(self):
        print("Car is Driving.")


# v = Vehicle()
# v.start()

# c = Car()
# c.drive()


c = Car()
c.start()  # start is the method of parent class
c.drive()  # drive is the method of child class
````

Note:
Single Inheritance is most basic and safest inheritance. It avoids Complixity and ambiguity.

[Ambiguity in Object-Oriented Programming (OOP) occurs when a compiler or program cannot decide which function, variable, or path to choose]    

### ★ 2. Multiple Inheritance
Meaning

In `Multiple Inheritance`,a child class inherits from more than one parent class.

📌 Structure

```text
    Parent1           Parent2
        \              /
         \            /
          \          /
             Child
```

### Why It is used?
 * When one class needs feature of multiple classes.
 * Helps to combine functionlity.

### Real-Life Example
 * Parent1: Camera
 * Parent2: Phone
 * Child: SmartPhone

`SmartPhone is both a Phone and a Camera`

Python Example

````python
class Camera: # Parent 1
    def click_photo(self):
        print("Photo Clicked......")


class Phone:  # Parent 2
    def calling(self):
        print("Calling......")


class SmartPhone(Camera,Phone): # Derived class : Parent1 + Parent2
    def internet(self):
        print("Google Connected")


obj = SmartPhone()
obj.click_photo()
obj.calling()
obj.internet()
````

Note:

Python support multiple inheritance, you can check order of method called Method Resoluction Order(MRO) to avoid confustion.



````python
print(SmartPhone.mro())
````

### ★ 3. Multilevel Inheritance

Meaning

In `Multilevel Inheritance`, inheritance happens in levels.


📌 Structure

`Grandparent ➽ Parent ➽ Child`

#### Why it is used?
 * To represent hierarchy in real world.
 * Step-by-step Inheritance of feature.

#### Real-Life Example

 * Grandparent: Animal
 * Parent: Mammal
 * Child: Dog

`Dog is a Mammal, Mammal is an Animal`

````python
class Animal:
    def eat(self):
        print("Eating......")

class Mammal(Animal):
    def walk(self):
        print("Walking")

class Dog(Mammal):
    def bark(self):
        print("Barking....")


d = Dog()
d.eat() # eat is the method of animal
d.walk() # walk is the method of mammal
d.bark() # bark is the mehtod of dog.
````

````python
print(Dog.mro())
````

### ★ 4.Hierarchical Inheritance

Meaning 

In `Hierarchical Inheritance,`, Multiple classes inherits from a single parent.


📌 Structure

```text

       Parent1
      /   |   \
     /    |    \
    /     |     \
 child1  child2  child3
 ```

### Why it is used?
* One parent has common functionality.
* Different children use and extend it differently.

#### Real-Life Example:
* Parent: Employee
* Children:
     * Manager
     * Developer
     * Tester
* All are employees but have different roles.

 Python Example

````python
class Employee:
    def work(self):
        print("Employee is working..........")

class Manager(Employee):
    def manage(self):
        print("Manager is Managing.......")

class Developer(Employee):
    def code(self):
        print("Developer is coding..........")
        
class Tester(Employee):
    def test(self):
        print("Tester is testing............")

m = Manager()
m.work()  # work() is the method of parent class
m.manage() # manage() is the method of Manager.

d = Developer()
d.work() # work() is the method of parent class.
d.code() # code() is the method of Developer class

t = Tester()
t.work() # work() is the method of parent class.
t.test() # test() is the method of Tester() class
````

````python
print(Manager.mro())
````

````python
print(Developer.mro())
````

````python
print(Tester.mro())
````

Note:
This is very common in software systems.
(Banking,College System,EMS)

#### ★ 5. Hybrid Inheritance

meaning

Hybrid Inheritance is combination of two or more inheritance type.

It may Include:
* Multiple+Hierarchical
* Hierarchical+Multilevel

```text
          A
        /   \
       B     C
        \   /
          D 
```
Here,
* B and C inherits from A(Hierarchcal Inheritance)
* D inherits from B and C(Multiple Inheritance)

### Why it is used?
* When a system requires complex Inheritance.
* Common in large framework and enterprises 

Python Example

````python
class A:
    def showA(self):
        print("Class A")
class B(A):
    def showB(self):
        print("Class B")
class C(A):
    def showC(self):
        print("Class C")
class D(B,C):
    def showD(self):
        print("Class D")


obj = D()
obj.showA()
obj.showB()
obj.showC()
obj.showD()
````

Note:

Hybrid Inheritance is powerful but can cause confusion

 * Method Conflict
 * ambiguity

````python
print(D.mro())
````
