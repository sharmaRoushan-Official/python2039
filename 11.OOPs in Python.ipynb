### 2. Class 
A `Class` is a blueprint or template used to create any object.

syntax

````python
class Student:
    pass
````

### Key Points:

* Define `class` keyword
* Can contains variables and method.
* Naming Case: PascalCase

````python
class Student:  # Student is the class name 
    def display(self):  # display is the method of Student Class
        print("Welcome to Python OOP") # Statment of method 
````

### 5. Constructor

A `Constructor` initializes object data autmatically when the objcet is created.


syntax:

`def __init__(self):`


````python
class Student:
    def __init__(self,name,age):
        self.name = name  # instance variable
        self.age = age  # instance varaible        
````

````python
s1 = Student("Ankesh",65)  # Function invoking
s1.name
````

````python
s1.age
````

### Key Points:
* Automatically invoded
* Used to initialized instance varaible

### Instance, Attribute and Method in Python

````python
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def studentDetails(self): # instance method 
        print("Student Name:",self.name)
        print("Student Age:",self.age)
obj = Student("Ankesh",13)
# print(obj.name)
# obj.age
obj.studentDetails()
````

````python
class Student:
    def __init__(ankesh,name,age):
        ankesh.name = name
        ankesh.age = age

    def studentDetails(ankesh): # instance method 
        print("Student Name:",ankesh.name)
        print("Student Age:",ankesh.age)
        
obj = Student("Ankesh",13)
# print(obj.name)
# obj.age
obj.studentDetails()
````
