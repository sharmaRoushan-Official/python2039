## 11. Class and Object in Python

#### 1. Object-Oriented Programming

#### Q. What is OOP?
`OOP` is a programming paradigm(An OOP paradigm is a style of coding that built software around object instead of logic) that organize software design around object, rather than function and logic.


#### Why OOP:

* Code Reusability
* Moudularity: [Moudlarity in OOP is the practice of breaking a complex software system into smaller, independent, and self-contained units.]
* Scalability: [Scalability in Object-Oriented Programming (OOP) means writing modular, flexible code that grows smoothly]
* Easy Maintainance
* Real-Wordl Modeling
