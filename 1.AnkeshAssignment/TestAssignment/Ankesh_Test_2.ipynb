```python
year = int(input("Enter the Year : "))

if year%400 == 0:
    print(f"{year} is leap year")
elif year%100 == 0:
    print(f"{year} is not leap year")
elif year % 4==0:
    print(f"{year} is leap year")
else:
    print(f"{year} is not leap year")
```

##### 5.  Check whether a number is an Armstrong number

```python
num = int(input("Enter a Number :" ))

digit =len(str(num))
temp = num
sum1 = 0
while temp>0:
    digits = temp%10
    sum1 = sum1+ digits**digit
    temp =temp//10
if num == sum1:
    print("Armstrong")
else:
    print("Not Armstrong")
```

##### 1.  Write a function to find the greatest of three numbers.


```python
def greatest(num1,num2,num3):
    if num1>num2 and num1>num3:
        print(f"{num1} is Largest ")
    elif num2>num3 and num2>num1:
        print(f"{num2} is Largest ")
    elif num3>num1 and num3>num2:
        print(f"{num3} is Largest ")
    else:
        print("All are Equal ")
        
greatest(30,40,20)
```

##### 2.  Write a function to check whether a number is prime.


```python
def prime(n):
    is_Prime=True

    if n<=1:
        print(f"{n} is not Prime")

    for i in range(2,n):
        if n%i == 0:
            is_Prime = False
     
    if is_Prime:
        print(f"{n} is Prime")
    else:
         print(f"{n} is not Prime")
    
n = int(input("Enter a Number :"))
prime(n)
    
```

## 4. Exception Handling



##### 1.  Write a program to handle ZeroDivisionError.


```python
try:
    a = int(input("Enter a Number 1"))
    b = int(input("Enter a Number 2"))

    print(a/b)
except ZeroDivisionError:
    print("Cannot Divide by Zero ")
except ValueError:
    print("Enter Valid Number")
```

##### 3.  Create a custom exception to check if age is less than 18.


```python
class AgeError (Exception):
    pass

try:
    age = int(input("Enter Your Age : "))
    if age<18:
        raise AgeError("Invalid Age, Please Enter age above 18 ")
        
except AgeError as e:
    print(e)
else:
    print("Eligible")
finally:
    print("Thank You")
```
