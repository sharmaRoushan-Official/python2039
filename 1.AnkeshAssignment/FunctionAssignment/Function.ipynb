##### 6. Write a function max_of_three(a, b, c) that returns the maximum of three numbers.

```python
def max_of_three(a,b,c):
    if a>b and a>c:
        return (f"{a} is greater than {b} and {c}")
    elif b>a and b>c:
        return  (f"{b} is greater than {a} and {c}")
    else:
        return (f"{c} is greater than {a} and {b}")
```

```python
print(max_of_three(3,9,6))
```

##### 7. Write a function factorial(n) that returns factorial of a number.

```python

def factorial(n):
    if n == 1 or n == 0:
        return 1
    else:
        return(n * factorial(n-1))
```

```python
 print(factorial(4))
```

##### 8. Write a function sum_list(lst) that returns the sum of all elements in a list.

 ##### 9. Write a function count_vowels(s) that counts vowels in a string.


```python
def count_vowels(s):
    count = 0
    vow ="aeiouAEIOU"
    for i in s:
        if i in vow:
            count +=  1
    return count
```

##### 10. Write a function reverse_string(s) that returns the reverse of the string.

```python
def reverse_strings(s):
    rev = ""
    for i in s:
        rev = i+rev
    return rev
        
```

```python
reverse_strings("Python")
```

##### 11. Write a function is_palindrome(s) that checks whether the string is palindrome or not.

```python
def is_palindrome(s):
    rev = ""
    for i in s:
        rev = i+rev
    if s == rev:
        print("palindrome")
    else:
        print(" not palindrome")
    
```

```python
is_palindrome("mada")
```

##### 15. Write a function count_words(sentence) that counts number of words in a sentence.
    

```python
def count_words(sentence):
    count = 1
    for i in sentence:
        if i ==" ":
            count += 1
    return count
```

```python
count_words("The Quick Brown Fox") 
```
