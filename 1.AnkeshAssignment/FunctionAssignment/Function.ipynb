##### 11. Write a function is_palindrome(s) that checks whether the string is palindrome or not.

```python
def is_palindrome(s):
    rev = ""
    for i in s:
        rev = i+rev
    if s == rev:
        print("palindrome")
    else:
        print(" not palindrome")
    
```

##### 18.Write a function second_largest(lst) that returns the second largest number.

```python
def  second_largest(lst):
    largest = lst[0]
    s_largest = lst[0]

    for i in lst:
        if i>largest:
            s_largest = largest
            largest = i
        elif i>s_largest and i!=largest:
            s_largest = i
    return (("second Largest is "),s_largest)

lst = [10,20,30,50,40,90,45]
second_largest(lst)
            
    
```

##### 19.Write a function prime_check(n) that returns True if number is prime.

```python
def prime_check(n):
    if n<=1:
        return False
        
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

num = int(input("Enter a number: "))
print(prime_check(num))
```

##### 20.Write a function generate_primes(n) that generates all prime numbers up to n.

```python
def generate_primes(n):

    for num in range(1, n + 1):

        if num <= 1:
            continue

        prime = True

        for i in range(2, num):
            if num % i == 0:
                prime = False
                break

        if prime:
            print(num, end=" ")

num = int(input("Enter the range: "))
generate_primes(num)
```

##### 23.Write a function merge_lists(list1, list2) that merges two lists without duplicates.


```python
def merge_lists(list1, list2):
    merged =[]
    for i in list1:
        if i not in merged:
            merged.append(i)

    for i in list2:
        if i not in merged:
            merged.append(i)
    print(f"List 1 is : {list1}")
    print(f"List 2 is : {list2}")
    print(f"Merged List is : {merged}")

list1 = [10,20,30,40,50,60,70,880]
list2 = [5,10,15,20,25,30,45,50,60,70,75]
merge_lists(list1, list2)
```
