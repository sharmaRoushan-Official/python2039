##### 1️⃣ Student Registration Validation

Validate:
* Name should contain only alphabets and at least 3 characters.
* Age should be between 18 and 30.
* Email should contain @ and .com.

Task:
If any validation fails, raise a Custom Exception showing all errors.

Sample Output:
Enter Name: R1
Enter Age: 16
Enter Email: abcgmail

Custom Exception:
Name Invalid, Age Invalid, Email Invalid


```python
class InvalidAgeError(Exception):
    pass
class InvalidNameError(Exception):
    pass
class InvalidEmailError(Exception):
    pass
```

```python
try:
    name = input("Enter your Name: ")
    if len(name)<3 or not name.isalpha():
        raise InvalidNameError("Invalid Name")

    age = int(input("Enter Your Age : "))
    if age<18 or age>30:
        raise InvalidAgeError("Enter Valid Age" )

    email = input("Enter Your Email : ")
    if "@" not in email or ".com" not in email:
        raise InvalidEmailError("Enter Valid Email")

except InvalidAgeError as e:
    print(e)
except InvalidNameError as e:
    print(e)
except InvalidEmailError as e:
   print(e)

else:
    print(f"Your name is {name} \nYour age is {age}\nYour email is {email}")
    print("Registration Successful")
finally:
    print("Thank You !")
        
```


##### 2️⃣ Employee Registration

Validate:
* Employee ID must start with "EMP".
* Salary should be greater than 20,000.
* Experience should be at least 1 year.

Task:
Raise one Custom Exception if validation fails.



```python
class InvalidIDError(Exception):
    pass
class InvalidSalError(Exception):
    pass
class InvalidExpError(Exception):
    pass
```

```python
try:
    empId = input("Enter Employee Id start with EMP : " )
    if not empId.startswith('EMP'):
        raise InvalidIDError ("Invalid Id !  ,  Employee Id Must Starts with EMP")
        
    sal=float(input("Enter Your Salary : "))
    if sal<20000:
        raise InvalidSalError("Invalid Salary ! ,  Salary should be greater than 20,000 ")
    
    exp = float(input("Enter your Experience in Year : "))
    if exp<1:
        raise InvalidExpError("Invalid Experience ! , Experience Should be at Least 1 year ")


except InvalidIDError as e:
    print(e)
except InvalidSalError as e:
    print(e)
except InvalidExpError as e:
    print(e)


else:
    print(f"Employee Id is {empId} \n Salary is {sal} \n Experience is {exp} \n")
    print("Registration Successful ")
finally:
    print("Thank You !!! ")
    
```

```python
class InvalidPinError(Exception):
    pass
class InvalidExceedError(Exception):
    pass
class InvalidMinWithdrawlError(Exception):
    pass
```

```python
try:
    balance = 20000
    
    pin = input("Enter Your Pin : ")
    if len(pin) != 4 or not pin.isdigit():
        raise InvalidPinError("Invalid Pin , Please enter 4 digit Pin only digits ")

    withdrawl = float(input("Enter the amount withdrawl amount :"))
    if withdrawl>balance:
        raise InvalidExceedError("Invalid withdrawl Amount , Withdrawl Amount should not be greater than Available Balance")
    elif withdrawl<100:
        raise InvalidMinWithdrawlError("Invalid Withdrawl Amount,  Withdrawl  amount should be greater than 100")
        

except InvalidPinError as e:
    print(e)
except InvalidExceedError as e:
    print(e)
except InvalidMinWithdrawlError as e:
    print(e)
else:
    print(f"\n withdrawl amount is {withdrawl} \n")
    print("Withdrawl Successfull \n")
finally:
    print("Thank You For Banking With us")
```

##### 4️⃣ Online Shopping Order

Validate:
* Product quantity must be greater than 0.
* Mobile number must contain exactly 10 digits.
* Delivery address cannot be empty.

Task:
Raise a Custom Exception displaying all validation errors.

━━━━━━━━━━━━━━━━━━━━

```python
class Invalid_Quant_Error (Exception):
    pass
class Invalid_M_number_Error(Exception):
    pass
class Invalid_Delv_Addr_Error(Exception):
    pass
```

```python
try:
    quant = int(input(" Enter the Number of Quantity : "))
    if quant<=0:
        raise Invalid_Quant_Error("Invalid Quantity , Quantity must be Greater than 0 ")

    mob = input("Enter Your Mobile Number : ")
    length = len(mob)
    if length != 10 or not mob.isdigit():
        raise Invalid_M_number_Error("Invalid Mobile Number , Mobile number should be 10 digit exactly")

    del_add = input("Enter the Delivery Address : ")
    if del_add.strip() =="":
        raise Invalid_Delv_Addr_Error("Invalid Delivery Address , Delivery Addree should not be Empty")





except Invalid_Quant_Error as e:
    print(e)
except Invalid_M_number_Error as e:
    print(e)
except Invalid_Delv_Addr_Error as e:
    print(e)

else:
    print(f"Quantity is {quant}\nMobile Number is {mob}\nDelivery Address is {del_add}\n")
    print("Order  Successfull\n ")
finally:
    print("Thank You ")
```
