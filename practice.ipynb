# 🐍 Python Practice Questions — Moderate Level

### 1. Find the Second Largest Number
Find the second largest number in a list without using `sort()`.

**Example:** `[10, 25, 8, 40, 15] → 25`

---

### 2. Remove Duplicates
Remove duplicate elements from a list without using `set()`, while maintaining the original order.

**Example:** `[1, 2, 2, 3, 4, 4, 5] → [1, 2, 3, 4, 5]`

---

### 3. Character Frequency
Count the frequency of each character in a string using a dictionary.

**Example:** `"programming" → {'p': 1, 'r': 2, ...}`

---

### 4. Word Frequency
Count how many times each word appears in a sentence.

**Example:** `"python is easy and python is powerful"`

**Output:** `python: 2, is: 2, easy: 1, and: 1, powerful: 1`

---

### 5. Palindrome Sentence
Check whether a sentence is a palindrome. Ignore spaces and capitalization.

**Example:** `"Madam" → Palindrome`

---

### 6. Anagram Checker
Check whether two strings are anagrams.

**Example:** `"listen"` and `"silent" → Anagram`

---

### 7. Prime Numbers in a Range
Take `start` and `end` values from the user and print all prime numbers within that range.

**Example:** `10–30 → 11, 13, 17, 19, 23, 29`

---

### 8. Find the Missing Number
A list contains numbers from `1` to `n`, but one number is missing. Find the missing number.

**Example:** `[1, 2, 3, 5, 6] → 4`

---

### 9. Find Common Elements
Find the common elements between two lists without using `set()`.

**Example:** `[1, 2, 3, 4]` and `[3, 4, 5, 6] → [3, 4]`

---

### 10. Rotate a List
Rotate a list to the right by `k` positions.

**Example:** `[1, 2, 3, 4, 5]`, `k = 2 → [4, 5, 1, 2, 3]`

---

### 11. Find Pairs with a Given Sum
Find all pairs of numbers whose sum is equal to a given target.

**Example:** `[2, 7, 4, 5, 3]`, target `9 → (2, 7), (4, 5)`

---

### 12. Flatten a Nested List
Convert a nested list into a single list.

**Example:** `[[1, 2], [3, 4], [5, 6]] → [1, 2, 3, 4, 5, 6]`

---

### 13. Student Marks Dictionary
Create a dictionary containing student names and their marks. Find:
- Student with the highest marks
- Student with the lowest marks
- Average marks

**Example:** `{"Rahul": 85, "Amit": 72, "Priya": 91, "Neha": 78}`

---

### 14. Sort a Dictionary by Value
Sort a dictionary based on its values in ascending order.

**Example:** `{"A": 50, "B": 20, "C": 80} → {"B": 20, "A": 50, "C": 80}`

---

### 15. Separate Even and Odd Numbers
Divide a list into two separate lists containing even and odd numbers.

**Example:** `[1, 2, 3, 4, 5, 6] → Even: [2, 4, 6], Odd: [1, 3, 5]`

---

### 16. Find the Longest Word
Find the longest word in a sentence. If multiple words have the same maximum length, print all of them.

**Example:** `"Python programming is interesting"`

**Output:** `programming, interesting`

---

### 17. Find Unique Elements
Count the frequency of numbers in a list and print only the numbers that appear exactly once.

**Example:** `[1, 2, 2, 3, 4, 4, 5] → [1, 3, 5]`

---

### 18. Password Validator
Create a function to validate a password. It must contain:
- At least 8 characters
- At least 1 uppercase letter
- At least 1 lowercase letter
- At least 1 digit
- At least 1 special character

**Example:** `"Python@123" → Valid Password`

---

### 19. ATM Simulation
Create a simple ATM program with the following options:

1. Check Balance
2. Deposit Money
3. Withdraw Money
4. Exit

Handle insufficient balance, invalid amounts, and invalid menu choices.

---

### 20. Mini Expense Tracker
Create a simple expense tracker where the user can enter:
- Expense name
- Amount
- Category

The program should:
1. Calculate total expenses
2. Find the highest expense
3. Calculate category-wise expenses
4. Display all expenses
