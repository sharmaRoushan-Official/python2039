```python
##Q1 creat a base class animal with a method speak().create two subclasses dog and cat that override speak() and print different sound.

#parent class
class Animal:
    def speak(self):
        print("animal sound")
        
#child class
class Dog:
    def speak(Animal):
        print("dog sound: WOOF WOOF")

class Cat:
    def speak(Animal):
        print("cat sound is:MEOW MEOw ")


#create object

dog=Dog()
cat=Cat()

# Calling methods
dog.speak()
cat.speak()
```
