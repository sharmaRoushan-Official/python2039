```python
#Q 16. print this pattern:

for i in range(5):
    for j in range(i):
        print("*",end=" ")
    print()
            
```

```python
#Q17 print multiplication table
num=int(input("enter a number:"))
for i in range(1,num+1):
    for j in range(1,11):
        print(f"{i}*{j}={i*j}")
        
        print()
    
```

```python
#Q20  print floyds triangle
n=int(input("enter a number:"))
num=0
for i in range(1,n+1):
    for j in range(1,i+1):
        num=num+1

        print(num,end=" ")
        
    print()
```
