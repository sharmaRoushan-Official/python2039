```python
## home work
#Q15
# creat a simple calculator that perform basic arithmetics operation based on two user point

a= float(input("enter  number a: "))
b= float(input("enter number b:"))
operator=input("enter operatot")

if operator=="+":
    print(a+b)
elif operator=="*":
    print(a*b)
elif operator=="%":
    print(a%b)
elif operator=="-":
    print(a-b)
elif operator=="**":
    print(a**b)
elif operator=="/":
    if a !=0:
        print(a/b)
    else:
        print("not divible by zero")

else:
    print("invalid number")


```
