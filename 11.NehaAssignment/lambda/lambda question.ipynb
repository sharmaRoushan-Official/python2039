LAMBDA PRACTICE QUESTION

```python
## write a lambda function to calculate the square of a number

x=float(input("enter a number"))
square= lambda x:x*x
print(square(x))
```

```python
#2 use a lambda function with map() to add 5 to each numberin the list:
nums=list(map(int,input("enter a list:").split()))
add=list(map(lambda x:x+5,nums))
print(add)
```

```python
#3 use filter() with lambda to find even numbers from a list.
nums=list(map(int,input("enter a list:").split()))
even=list(map(lambda x:x%2==0, nums))
print(even)
```

```python
#4 use reduce() with a lambda to compute the product of all element in a list:
from functools import reduce
nums=list(map(int, input("enter a number:").split()))
product=reduce(lambda a,b:a*b, nums)
print(product)
```

```python
#5 short the following list of tuples by the second value using a lambda function.

nums= [(1,2),(2,1),(4,2)]
short=lambda x:x.short(),nums
print(short)
```

```python
#6 write a lambda that check if a string starts with 'A'

x=["Ankit","Rahul","Umesh","Ankesh"]
string=list(filter(lambda x:x.startswith("A"),x))
print(string)
```

```python

nums=input("enter a string:").split()
string=list(filter(lambda x: "A" in x, nums))
print(string)
```

```python
#7 use a lambda function with map() to convert a list of celsius temprature to fahrenheit:

#temp=[0,20,37,100]
temp=list(map(int,input("enter a number:").split()))
calculate=list(map(lambda C: (C*9/5)+32, temp))
print(calculate)
```

```python
#8 given a list of strings filter all strings with length>3 using a lambda

words=['pen', 'notebook', 'apple', 'go']
li=list(filter(lambda x:len(x)>3,words))
print(li)
```
