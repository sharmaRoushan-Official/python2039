list comprehention

syntax:- new_list[expression for item in iterable if condition]

syntax:-new_value if condition else old_value for x in list

```python
#Q1 generate a list of square from 1 to 10 using list comprehension

square=[x**2 for x in range(1,10)]
print(square)
```

```python
#q2 create a list of even numbers from  1 to 50 using list comprehenstion
#x=int(input("enter a number:"))
even=[x for x in range(1,50) if x%2==0]
print(even)
```

```python
#4 convert the list of string to uppercase using list comprehension: words=['cat', 'dog', 'lion']

word=[x.upper() for x in ['cat', 'dog', 'lion'] ]
print(word)
```

```python
#5 get a list of numbers divisible by both 3 and 5 from 1 to 100

number=[x for x in range(1,100) if x%3==0 and x%5==0]
print(number)

```

```python
#Q6 Extract all vowels from string using list comprehension text='comrehension'

text='comprehension'
string=[ x for x  in text if  x in 'aeiou']
print(string)
```

```python
#Q7 given a list of string of words extract words that contain the letter'e':
words=['tree','book','python', 'sky']
string=[x for x in words if  "e"  in x ]
print(string)
```

```python
#q8 create a list of all 3 digit numbers that are palindromes using list comprehensiomn

number=[x for x in range(100,1000) if (x//100)==(x%10)]
print(number)
```

```python
#9 from a list of number replace odd and keep even number as is:
nums=[1,2,3,4,5,6,]
odd=["odd" if x%2!=0 else x for x in nums]
print(odd)
```
