```python
#Q5 write a program to calculate the electricity bill.

# first 100 units- 5 unit
# next 100 unit -7 unit
# above 200 unit - 10 unit

unit= float(input("enter a number:"))
bill=0
if unit>=100:
    bill=unit*5      
elif bill<=200:
    bill=(100*5)+((unit-100)*7)
elif bill>200:
    bill=(100*5)+(100*7)+((unit-200)*10)
print(bill,": electricity bill")
```

```python
# check whether a number is an armstrong number.
num=int(input("enter a number:"))
digit=len(str(num))
total=0
for i in str(num):
    total=total+int(i)**digit
if total==num:
    print("number is armstrong")
else:
    print("number is not armstrong")
print(total)
```

```python
#Q5 write a function  that accepts a list and return the second largest element.
def number(nums):
    nums.sort()
    return nums[-2]
    
numbers=list(map(int,input("enter a number:").split(",")))
print("second_largest",number(numbers))
```

```python
#Q3 write a function that return the reverse string

def reverse(st):
    return st[::-1]

text=str(input("enter a text:"))
print("reverse string",reverse(text))
```

```python
#Q2 write a functo check prime number

def prime(n):
   # n=int(input("enter a number:"))

    if n<=1:
        print(" not prime")
    else:
        for i in range(2,int(n**0.5)+1):
            
            if n%i==0:
                print("is not prime")
                break
        
        else:
            print("is prime")

            

num=float(input("enter a number:"))
print("prime ",prime(num))
    
```

```python
#Q1 write a program to handle zero division error

try:
    num1=float(input("enter a number num1:"))
    num2=float(input("enter a number num2:"))
    
    print(num1/num2)
    
except ZeroDivisionError:
    print("cannot divide by zero:")
    
```

```python
#Q2 write a program handle value error when the user enter a invalid input

try:
    num1=int(input("enter a number:"))
    num2=int(input("enter a number:"))
    print(num1*num2)
except ValueError:
    print("user enter a invalid input")
```
