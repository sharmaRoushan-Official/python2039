Note:

* if condition is True, code runs

* If False, nothing happens

* Else block runs directly when conditions is False.

* Either if block or Else will be execute --- never both

### 7. if-elif-else (Multiple Conditions)


syntax:
-------

if condition1:
    
    statement1

elif condition2:
    
    statement2

elif condition3:
    
    statment3

else:
    statement4

```python
marks = 95

if marks>=90:
    print("Grade A")
elif marks>= 75:
    print("Grade B")
elif marks>= 60:
    print("Grade C")
else:
    print("Fail")
```

Flow Rule:

* Python check top to bottom
* First True condition Execute
* Remaining Condition are skipped.

### 8. Nested if (condition inside another condition)

syntax:

if codition:

    if condition:
    
        statement

```python
age = 15

citizen = True

if age>=18:   # True
    # print("Elegible to vote")   # True Block 
    if citizen: # True
        print("Elegible to Vote")
    else:
        print("You are not citizen of India")
else:
    print("Not Elegible to vote.")
```

### 9. Short Hand if (One-line condtion)

statement if codition

```python
age = 15


"Elegible to Vote" if age>=18 else "Not Elegible to vote"
```

```python
# 21. Write a Python program to calculate *discount* based on purchase amount:

#     * > 1000 → 20% discount
#     * 500–1000 → 10% discount
#     * <500 → No discount


amount = int(input("Enter your shopping amount ₹: "))

if amount>0:
    if amount>=1000:
        dis = "20%"
        disAmount = (amount*20)/100
        # print(disAmount)
        totalAmount = amount-disAmount
        # print(totalAmount)
        print("===========Shopping Receipt==============")
        print(f"Total Shopping Amount ₹:{amount}\nTotal Discount ₹:{dis}\nTotal discount Amount ₹: {disAmount}\nTotal Amount to pay ₹:{totalAmount}")
    elif amount>=500 and amount<1000:
        dis = "10%"
        disAmount = (amount*10)/100
        # print(disAmount)
        totalAmount = amount-disAmount
        # print(totalAmount)
        print("===========Shopping Receipt==============")
        print(f"Total Shopping Amount ₹:{amount}\nTotal Discount ₹:{dis}\nTotal discount Amount ₹: {disAmount}\nTotal Amount to pay ₹:{totalAmount}")
        getDiscount = 1000 - amount
        print(f"Warning: If you Shopping more amount ₹:{getDiscount} then you will get 20% Discount")
    else:
        dis = "0%"
        disAmount = (amount*0)/100
        # print(disAmount)
        totalAmount = amount-disAmount
        # print(totalAmount)
        print("===========Shopping Receipt==============")
        print(f"Total Shopping Amount ₹:{amount}\nTotal Discount ₹:{dis}\nTotal discount Amount ₹: {disAmount}\nTotal Amount to pay ₹:{totalAmount}")
        getDiscount = 500 - amount
        print(f"Warning: If you Shopping more amount ₹:{getDiscount} then you will get 10% Discount")
else:
    print(f"{amount} is not valid or Enter Positive value")
```

```python
# 30. Write a Python program to check whether an input string is *Number, **only spaces, or **contains characters*, **special symbol.


value = input("Enter any value: ")

if len(value)<2:
    if ord(value) == 32:
        print(f"You entered space {value}")
    elif ord(value)>64 and  ord(value)<91:
        print(f"You Entered capital character: {value}")
    elif ord(value)>96 and ord(value)<123:
        print(f"You Entered small character: {value}")
    elif ord(value)>32 and ord(value)<48 or ord(value)>57 and ord(value)<65:
        print(f"You Entered Special Character: {value}")
    elif ord(value)>47 and ord(value)<58:
        print(f"You Entered Numeric Character: {value}")
else:
    print(f"You Entered more than one character:{value} Invalid Character")
```

```python
ascii_lowercase = 'abcdefghijklmnopqrstuvwxyz'
ascii_uppercase = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
digits = '0123456789'
punctuation = '!"#$%&\'()*+,-./:;<=>?@[\\]^_`{|}~'
whitespace = ' \t\n\r\x0b\x0c'

value = input("Enter any character: ")

if len(value) == 1:
    if value in ascii_lowercase:
        print(f"You Entered Lower Character: {value}")
    elif value in ascii_uppercase:
        print(f"You entered Upper Character: {value}")
    elif value in digits:
        print(f"You Entered digits: {value}")
    elif value in punctuation:
        print(f"You Entered special character: {value}")
    elif value in whitespace:
        print(f"You entered white space: {value}")
else:
    print("Invalid Character length.")
```
