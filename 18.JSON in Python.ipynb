#### Working with JSON in Python

-----
Python provides built-in module:

`import json`

```python
import json
```

### 𝟏. Convert Python -> JSON

json.dumps() [Dictionary to JSON String]

```python
data = {
    "name":"Ram",
    'age':25,
    "city":"Delhi"
}
print(data,type(data))
```

```python
json_data = json.dumps(data)
print(json_data,type(json_data))
```

```python
json_data = json.dumps(data,indent=4)
print(json_data)
```

Note:

Converts dictionary to JSON string

### 𝟐.Convert JSON -> Python

##### json.loads() [JSON string to dictionary]

```python
json_data1 = '{"name": "Ram", "age": 25, "city": "Delhi"}'
print(json_data1)
```

```python
data = json.loads(json_data1)
print(data)
print(type(data))
```

```python
data['name'] = "Ankesh"
data
```

```python
json_string = json.dumps(data,indent=4)
print(json_string)
```

#### 𝟑. Working with JSON File 

##### Writing JSON to File
`json.dump() -> Python Object  -  File`

```python
data = {
    "name": "Ankesh",
    "age": 25,
    "city": "Delhi"
}


with open("student.json",'w') as file:
    json.dump(data,file,indent=4)
    
```

dump() -> Write to file

```python
with open("student1.json",'w') as file:
    json.dump(data,file,indent=4)
```

### Reading JSON from File 

```python
with open("student.json",'r') as file:
    data = json.load(file)
    print(data)
```

load() -> Read from file

##### dump() vs dumps() and load() vs loads()
