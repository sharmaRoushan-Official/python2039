### 9. The else Block

When does else execute?

Executes only if No exception occurs

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    div = a/b
except ZeroDivisionError:
    print("Cannot divide by Zero")
except ValueError:
    print("Please enter a valid number")
else:
    print("Result:",div)
```

### 10. The finally Block 

When does finally execute?

* Executes always
* Used for cleanup (closing files,DB connection)

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    div = a/b
except ZeroDivisionError:
    print("Cannot divide by Zero")
except ValueError:
    print("Please enter a valid number")
else:
    print("Result:",div)
finally:
    print("Code Executed Successfully")
```

```python
try:
    file = open("C:\\Users\\User\\Desktop\\NursaryRhymes.txt",'r')
except FileNotFoundError:
    print("File Name is Invalid! Please Enter a valid file Name.")
else:
    print("Output is:\n",file.read())
finally:
    print("Code Executed Successfully")
```

### 14. Custome Exceptions (User-Defined Exception)

You can create your own Exception class.

```python
class InvalidAgeError(Exception):
    pass
```

```python
age = int(input("Enter your age: "))

if age<18:
    raise InvalidAgeError("Age must be 18 or above")
else:
    print("Your are Elegible")
```

### Custom Excetpion : Age Invalid, Name Invalid, Email Invalid

```python
# name =
# age = 
# Email=
```

```python
class InvalidNameError(Exception):
    pass

class InvalidAgeError(Exception):
    pass

class InvalidEmailError(Exception):
    pass
```

```python
try:
    name = input("Enter your name: ")
    if len(name)<3 and not name.isalpha():
        raise InvalidNameError("Invalid Name Error")
        
    age = int(input("Enter your age: "))
    if age<18:
        raise InvalidAgeError("Invalid Age")
        
    email = input("Enter your email: ")
    if "@" not in email:
        raise InvalidEmailError("Invalid Email")

except InvalidNameError as e:
    print(e)
except InvalidAgeError as e:
    print(e)
except InvalidEmailError as e:
    print(e)
else:
    print("Your Name:",name,"\nYour Age:",age,"\nYour Email:",email)
    print("Registration Successful")
finally:
    print("Thank You!")
```
