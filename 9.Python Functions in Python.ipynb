### 7. Scope of Variables

1. Local Varaible
2. Global Variable

````python
def test():  # nullable function
    x1 = 10 # local Scope Variable
    print(x1)

    print(x1)
test()    
````

````python
print(x1)
````

### 2. Global Scope



````python
x2 = 100

def test():
    x3 = 45  # local Variable
    print(x3)
    print(x2) # calling inside the function,
test()
````

````python
print(x2) # calling outise the function 
````

````python
print(x3)
````

### global Keyword 

````python
x2 = 100  # global keyword

def test():
    global x3  # created globally
    x3 = 45  # local Variable
    print(x3)
    print(x2) # calling inside the function,
test()
````

````python
print(x2)
````

````python
print(x3)
````

### 8. Recursion:
Function Calling itself

````python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n-1)
factorial(4)
````

### 9. Nested Functions

Function Inside function

````python
def outerFunc(name):  # outer function
    print("Congratulations!"+name)
    def innerFunc():  # inner function
        name1 = input("Enter your name here: ")
        print("Hello!"+name1)
    innerFunc()
outerFunc("Ankesh")
````

### 17.Generator Function


````python
def count_up_to(n):
    for i in range(1,n+1):
        yield i
for num in count_up_to(5):
    print(num)
````

````python
x = count_up_to(5)
print(x)
````
