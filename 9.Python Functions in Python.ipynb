### 1. What is function?
A `function` is a resuable block of code that perform specific task.

#### Why we use function?

* Avoid code repetion
* Make code reuseable
* Improve Modularity
* Easy Debugging


### 2. Creating (Defining) a funciton

syntax:

```text
def function_name():
    # block of code
    

````python
def greet():  # greet() --> name of function (indentifier)
    print("Hello, Welcome to Python")
````

````python
greet()  # calling of function
````

#### 3. Function with Parameter(Arguments):

Parameters:

Values passed into a function 

````python
def greet(name): # def is keyword, greet() is the name of function, name --> variable is called parameter
    print("Congratulation "+name)
````

````python
greet("Swadhin")  # Swadhin --> Argument
````

### 5. Types of Function Arguments

1. Positional Arguments
2. Keyword Arguments
3. Default Arguments
4. Variable Length Arguments(*age) (**kwargs)

## 1.Positional Arguments

````python
def student(name,age):
    # print(name,age)
    return name,age
````

````python
obj = student("Satyadev",95)
obj
````

````python
print(obj)
````

````python
# positional arguemnt
def student(name,age):
    print(f"Student name is:{name}")
    print(f"Student age is:{age}")
    # print(name,age)
````

````python
student('satyadev',10)   # Correct position
````

````python
student(5,"Satyadev") # Wrong Position
````

````python
student("Satyadev")
````

### 2. Keyword Argument



````python
def student(name,age):
    print(f"Student name is:{name}")
    print(f"Student age is:{age}")


student(age=10,name="Swadhin")  # keyword Argument
````

````python
student(name="Swadhin",age=50)  # keyword argument
````
