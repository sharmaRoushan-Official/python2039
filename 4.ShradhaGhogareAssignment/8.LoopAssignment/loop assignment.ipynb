### 27. check if a string is a palindrome using loop.

```python
s = input("Enter a string:")
reverse = ""
for i in s:
    reverse = i + reverse
if s == reverse:
    print(s, "is a palindrome")
else:
    print(s, "is not a palindrome")
```

### 29. find the GCD of two numbers using a loop.

```python
def gcd(a, b):
    while b != 0:
        a , b = b, a % b
    return a
    
print(gcd(12, 18))
print(gcd(54, 24))
print(gcd(7, 9))
```
