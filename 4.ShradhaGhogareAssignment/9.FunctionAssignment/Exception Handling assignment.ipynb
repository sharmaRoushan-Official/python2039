### Q 1.Student Registration Validation

Validate:
* Name should contain only alphabets and at least 3 characters.
* Age should be between 18 and 30.
* Email should contain @ and .com.

Task:
If any validation fails, raise a Custom Exception showing all errors.

Sample Output:
* Enter Name: R1
* Enter Age: 16
* Enter Email: abcgmail

Custom Exception:
Name Invalid, Age Invalid, Email Invalid



```python
class InvalidNameError(Exception):
    pass
    
class InvalidAgeError(Exception):
    pass
    
class InvalidEmailError(Exception):
    pass
```

```python
try:
    name = input("Enter your name:")
    if len(name)<3 and not name.isalpha():
        raise InvalidNameError("invalid name")
        
    age = int(input("Enter your age:"))
    if age<=18 or  age>=30:
        raise InvalidAgeError("invalid age")
        
    email = input("Enter your email:")
    if "@" not in email:
        raise InvalidEmailError("invalid email")
        
except InvalidNameError as e:
    print(e)
except InvalidAgeError as e:
    print(e)
except InvalidEmailError as e:
    print(e)
else:
    print("Registration Successful")
finally:
    print("Thank you")
```
