### 10. Abstraction (Bank Example)
Create an abstract class Bank with abstract method rate_of_interest().

Create classes SBI and HDFC implementing different interest rates.
* Goal: Real-world abstraction usage.

```python
from abc import ABC, abstractmethod

class Bank (ABC):
    @abstractmethod
    def rate_of_interest(self):
        pass
        
class SBI(Bank):
    def rate_of_interest(self):
        return  6.6
        
class HDFC(Bank):
    def rate_of_interest(self):
         return  8.5
sbi = SBI()
hdfc = HDFC()
print ("SBI Rate of Interest:", sbi.rate_of_interest(), "%")
print ("HDFC Rate of Intersest:", hdfc.rate_of_interest(), "%")
```

### 1. Create a base class Animal with a method speak(). Create two subclasses Dog and Cat that override speak() and print different sounds.

### 2.Create a class Vehicle with attributes brand and model. Create subclasses Car and Bike that add a method vehicle_type() returning the type of vehicle.

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
        print("vehicle is created")
class Car(Vehicle):
    def vehicle_type(self):
        print("Car Brand Name:", self.brand)
        print("Car Model Name:", self.model)
class Bike(Vehicle):
    def vehicle_type(self):
        print("Bike Brand Name:", self.brand)
        print("Bike Model Name:", self.model)
obj = Car("Maruti suzuki", "Swift")
obj.vehicle_type()
obj1 = Bike("Royal Enfield", "classic 350")
obj1.vehicle_type()
```

### 3.Create a class Person with name and age. Create a derived class Student that adds student ID and grade. Write a method to print complete details.

```python
class Person:
    def __init__(self, name , age):
        self.name = name
        self.age = age
        print("person class is created")
    def display_details(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        
class Student(Person):
    def __init__(self,name, age, student_id, grade):
        super().__init__(name,age)
        self.student_id = student_id
        self.grade =  grade
        print(f"student class is created")
    def display_details(self):
        super().display_details()
        print(f"student ID: {self.student_id}")
        print(f"grade: {self.grade}")

obj =  Student("Rahul",29, "AZ10", "A+")
print("Details:-")
obj.display_details()
        
```
