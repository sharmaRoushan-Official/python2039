## Lambda Function

### 1. Write a lambda function to calculate the square of a number.

```python
sqrNum =(lambda a:a**2)
sqrNum(10)
```

### 2. Use a lambda function with map() to add 5 to each number in the list:
   nums = [1, 2, 3, 4]

```python
nums = [1, 2, 3, 4]
result =list(map(lambda a:a + 5, nums))
print(result)
```

### 3. Use filter() with a lambda to find even numbers from a list:
   nums = [5, 8, 11, 14, 17, 20]

```python
nums = [5, 8, 11, 14, 17, 20]
list(filter(lambda a: a % 2 ==0, nums))
```

### 4. Use reduce() with a lambda to compute the product of all elements in a list:
   nums = [2, 3, 4, 5]

```python
from functools import reduce
```

```python
nums = [2, 3, 4, 5]
product = reduce(lambda x, y: x * y, nums)
print(product)
```

### 5. Sort the following list of tuples by the second value using a lambda function:
   data = [(1, 3), (2, 1), (4, 2)]

```python
data = [(1, 3), (2, 1), (4, 2)]
new_data =sorted(data,key=lambda x: x[1])
print(new_data)
```

### 6. Write a lambda that checks if a string starts with 'A'.

```python
words =['Apple', 'Banana', 'Avocado','mango']
result = list(filter(lambda s: s.startswith('A'), words))
print(result)
```

### 7. Use a lambda function with map() to convert a list of Celsius temperatures to Fahrenheit:
   temps = [0, 20, 37, 100]

```python
temps = [0, 20, 37, 100]
fahrenheit = list(map(lambda c:(c *9/5 + 32), temps))
print(fahrenheit)
                  
                
```

### 8. Given a list of strings, filter all strings with length > 3 using a lambda:
   words = ['pen', 'notebook', 'apple', 'go']

```python
words = ['pen', 'notebook', 'apple', 'go']
result = list(filter(lambda x: len(x) > 3,words)) #more than 3
print(result)
```

### 9. Create a lambda that takes two arguments and returns their average.

```python
avg = lambda a, b: (a + b) / 2
print(avg(10,20))
```

### 10. Use a lambda and map() to capitalize the first letter of each word in a list:
    words = ['python', 'lambda', 'function']

```python
words = ['python', 'lambda', 'function']
result = list(map(lambda x: x.capitalize(),words))
print(result)
```

## List Comprehension

### 11. Generate a list of squares from 1 to 10 using list comprehension.

```python
squares = [i**2 for i in range(1,11)]
print(squares)
```

### 12. Create a list of even numbers from 1 to 50 using list comprehension.

```python
evens = [x for x in range(1,51) if x % 2 == 0]
print(evens)
```

### 13. Convert the list of strings to uppercase using list comprehension:
    words = ['cat', 'dog', 'lion']

```python
words = ['cat', 'dog', 'lion']
result = list(map(lambda x: x.upper(),words))
print(result)
```

### 14. Get a list of numbers divisible by both 3 and 5 from 1 to 100.

```python
nums = [x for x in range(1,101) if x % 3 == 0 and x % 5 == 0]
print(nums)
```

### 15. Extract all vowels from the string using list comprehension:
    text = "Comprehension"

```python
text = "Comprehension"
vowels = [c for c in text if c.lower() in 'aeiou']
print(vowels)
```

### 16. Given a matrix:
    matrix = [[1,2,3],[4,5,6],[7,8,9]]
    Use list comprehension to flatten it into a single list.

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
flat = [b for a in matrix for b in a]
print(flat)
```

### 17. Given a list of words, extract words that contain the letter ‘e’:
    words = ['tree', 'book', 'python', 'sky']


```python
words = ['tree', 'book', 'python', 'sky']
result =[x for x in words if 'e' in x]
print(result)
```

### 18. Create a list of all 3-digit numbers that are palindromes using list comprehension.


```python
palindromes = [x for x in range(100,1000) if str(x) == str(x)[::-1]]
print(palindromes)
```

### 19. From a list of numbers, replace odd numbers with 'ODD' and keep even numbers as is:
    nums = [1, 2, 3, 4, 5, 6]

```python
nums = [1, 2, 3, 4, 5, 6]
['ODD' if x % 2 != 0 else x for x in nums]
```

### 20. Using nested list comprehension, generate a 3x3 identity matrix.

```python
matrix = [[1 if i == j else 0 for j in range(3)] for i in range(3)]
print(matrix)
```
