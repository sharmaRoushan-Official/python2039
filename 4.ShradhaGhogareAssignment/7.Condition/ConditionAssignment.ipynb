### 21. write a python program to calculate *Discount* based on purches amount.
   * >1000 → 20 % discount
   * 500-1000 → 10 % discount
   * < 500 → No discount

```python
amount = float(input("Enter purches amount:"))

if amount >= 1000:
    discount = amount * 0.20
    print("Discount: 20%")
elif amount >= 500 and amount < 1000:
    discount = amount * 0.10
    print ("Discount: 10%")
else:
    discount = 0 
    print("Discount: No discount")
final_amount = amount - discount
print("Discount Amount: ₹", discount)
print("Final Amount to pay: ₹", final_amount)
```

### 25. write a python program to calculate the * electricity bill*:
