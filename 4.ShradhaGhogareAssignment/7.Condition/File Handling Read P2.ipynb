### Q 11. Read a file and count how many times a given word appears.


```python
fileName = r"C:\\Users\\ADMIN\\Desktop\\twinkle.txt"
word_to = input("Enter the word to search:").lower()
count = 0
with open(fileName, "r") as file:
    for line in file:
        words = line.lower().split()
        for word in words:
            word = word.strip('.,!?()]\"\'')
            if word == word_to:
                count += 1
print(f"'{word_to}' appears {count} times in the file")
```

### Q 19. Read a file and count the number of digits.

### Q 22. Read a file and display the top 10 most frequently used words.


```python
fileName =  r"C:\\Users\\ADMIN\\Desktop\\twinkle.txt"
with open(fileName,"r") as file:
    words = file.read().split()
    seen=[]
    count = 0
    for word in words:
        if word not in seen:
            seen.append(word)
            print(word, ":", words.count(word))
            count +=1
            if count == 10:
                break
            
        
    
    
```
