### Q 17. Write all prime numbers between 1 and 100 into a file.


```python
with open("PrimeNo.txt",'w') as file:
    for i in range(2,100):
        for j in range(2,i):
            if i % j == 0:
                break
        else:
            file.write(str(i)+'\n')
```

### Q 19. Write the first 20 Fibonacci numbers into a file.

```python
with open("FiboncciNo.txt",'w') as file:
    a , b = 0 , 1
    for i in range(20):
        file.write(str(a) +'\n')
        a, b = b, a + b
    
```

### Q 20. Create a file containing multiplication tables from 1 to 10.


```python
with open("MultiplicationT.txt",'w') as file:
    for i in range(1,11):
        file.write(f"\nTable of {i}\n")
        for j in range(1,11):
            file.write(f"{i} x {j} = {i*j}\n")
```

### Q 21. Take details of 5 students and store them in a file in this format:

Name, Roll No, Marks
