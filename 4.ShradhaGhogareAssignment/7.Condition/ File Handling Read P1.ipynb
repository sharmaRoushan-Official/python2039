### Q 28. Read a file and create a frequency dictionary of all words.

```python
fileName = r"C:\Users\ADMIN\Desktop\twinkle.txt"
word1 = {}
with open (fileName, "r") as file:
    for line in file:
        words = line.lower().split()
        for word in words:
            word = word.strip('.,()[]"\'')
            if word in word1:
                word1[word] += 1
            else:
                word1[word] = 1
                
print(word1)
print("\n-------Top 5 words-------")
for word, count in sorted (word1.items(), key = lambda x: x[1], reverse= True)[:5]:
    print(f"{word}: {count}")
```

### Q 29. Read a text file and determine whether it is a palindrome after removing spaces, punctuation, and case differences.


```python
fileName =  r"C:\\Users\\ADMIN\\Desktop\\twinkle.txt"

with open(fileName,"r") as file:
    
    text = file.read().lower()
    clean = ""
    for char in text:
        if char.isalnum():
            clean += char
            
      if clean == clean[::-1]:
          print("File content is palindrome")
      else:
          print("File content is NOT palindrome")
```
