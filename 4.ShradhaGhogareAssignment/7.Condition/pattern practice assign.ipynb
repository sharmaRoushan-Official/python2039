## Pattern Practice

### 1. Left Triangle
<pre>
*
**
***
****
*****
    
</pre>



```python
for i in range(1,6):
    for j in range(1, i+1):
        print('*',end='')
    print()
```

```python
for i in range(1,6):
    print('*' *i)
```

### 2. Reverse Left Triangle
<pre>
*****
****
***
**
*
</pre>

```python
for i in range(5,0,-1):
    for j in range(1, i+1):
        print('*',end='')
    print()
```

### 4. Right Triangle
<pre>
    *
   **
  ***
 ****
*****

</pre>
   

```python
for i in range(1,6):
    for j in range(5 - i):
        print(' ',end='')
    for k in range(i):
        print('*',end='')
    print()
    
```

### 5. Reverse Right Triangle
<pre>
*****
 ****
  ***
   **
    *
    
</pre>


```python
for i in range(5,0,-1):
    for j in range (5 - i):
        print(' ',end='')
    for k in range(i):
        print('*',end='')
    print()
```

### 6. Increment Number Triangle
 1<br>
 12<br>
 123<br>
 1234<br>
 12345

```python
for i in range(1,6):
    for j in range(1, i+1):
        print(j,end='')
    print()
```

### 7. Repeated Number Triangle
1<br>
22<br>
333<br>
4444<br>
55555

```python
for i in range(1,6):
    for j in range(i):
        print(i,end='')
    print()
```

### 8. Continuous Number Triangle
1<br>
2 3<br>
4 5 6<br>
7 8 9 10<br>
11 12 13 14 15

```python
num = 1
for i in range(1,6):
    for j in range(i):
        print(num, end='')
        num +=1
    print()
```

### 10. Alphabet Triangle
A<br>
AB<br>
ABC<br>
ABCD<br>
ABCDE

```python
for i in range(1,6):
    for j in range(1, i+1):
        print(chr(64 + j), end='')
    print()
```

### 11. Repeated Alphabet Triangle
A<br>
BB<br>
CCC<br>
DDDD<br>
EEEEE

```python
for i in range(1,6):
    char = chr(64 + i)
    print(char * i)
```

### 12. Reverse Alphabet Triangle
ABCDE<br>

ABC<br>

```python
for i in range(5,0, -1):
    for j in range(1, i+1):
        print(chr(64 + j), end='')
    print()
```
