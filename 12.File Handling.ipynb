### 12. Reading Files in Python

Q. What is File Handling?

`File Handling` allows python program to read data files and write data to files permanently.

* 📌 Why we read files:
   * Stored data permently
   * Read logs, reports, csv(Comma Seperatred value)
   * Process large data
   * Avoid manual input

### Types of Files
1. Text Files: [.txt,.csv,.log]
2. Binary Files: [.jpg,.pdf,.exe]

>  Here we focus on reading text files only.

### Reading File Using Loop

Best method for large files

```python
# step 1:

file = open(r"C:\Users\User\Desktop\NursaryRhymes.txt","r")

for line in file:
    print(line)
file.close()
```

```python
# step 1:

file = open(r"C:\Users\User\Desktop\NursaryRhymes.txt","r")

for line in file:
    for char in line:
        print(char)
file.close()
```

### 10. Using With Statemenet [Best Practice]

#### Checking File exists or Not 

`import OS Module`

```python
import os
```

```python
if os.path.exists(r"C:\Users\User\Desktop\NursaryRhymes.txt"):
    print("File Exists")
else:
    print("File Not Exists")
```

```python
fileName = r"C:\Users\User\Desktop\NursaryRhymes.txt"
if os.path.exists(fileName):
    print("File Exists")
    with open(fileName,'r') as f:
        print(f.read())
else:
    print("File Not Exists")
```
