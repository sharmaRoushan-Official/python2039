# 17. Generator in Python

#### What is a Generator?

A generator is a special function that returns an iterator and produces values one at a time using yield instead of return.

📌 Define: 
> A generator generates values one by one, only when needed.

-----

#### Why Generator Ared Needed (Problem ➽ Solution)


❌ Without Generator(List)

```python
def get_numbers():
    return [1,2,3,4,5]
get_numbers()
```

### ✅ With Generator

```python
list1 = [1, 2, 3, 4, 5]

def get_numbers():
    for num in list1:
        yield num
obj = get_numbers()
obj
```

```python
print(next(obj))
```

```python
print(next(obj))
```

```python
print(next(obj))
```

```python
print(next(obj))
```

```python
print(next(obj))
```

```python
print(next(obj))
```

```python
for value in get_numbers():
    print(value)
```

Notes:

* ꪜValues created on demand
* ꪜMemory efficient.

### yield VS return 

|return|yield|
|-----|----|
|Ends of Function|Pauses function|
|Return all value|Returns one-by-one values|
|Memory heavy|Memory Efficient|

### Generator with Loop

```python
def count_up(n):
    for i in range(1,n+1):
        yield i
```

### How Generator Works

📌 Steps

1. Function called -> no execution
2. Return Generator Object
3. next() resumes execution
5. Stop at yield 
6. Continues from same point

### Generator VS Normal Function

|Feature|Generator|Function|
|----|----|----|
|Uses|yield|return|
|Momory|Less|High|
|Execution|Lazy|Immediate|
|Values|One at a time|All at once|

## Real-World Example : Reading Files 

```python
def read_file(filename):
    with open(filename) as file:
        for line in file:
            yield line.strip()
```
