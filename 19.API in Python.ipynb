### 1. GET Request Example

```python
import requests
response = requests.get("https://jsonplaceholder.typicode.com/posts")

print(response.status_code)  # Congratulations
print(response.json())
```

## 2. POST Request Example

```python
import requests
```

```python
data = {
    "title":"New Post",
    "body":"This is an API Testing",
    "userID":1
}


response = requests.post("https://jsonplaceholder.typicode.com/posts",json=data)
print(response.status_code)
print(response.json())
```

### 3. PUT Request Example 


```python
data = {
    "title":"PM of INDIA",
    "body":"I am Currenlty PM of INDIA",
    "id":101
}


response = requests.put("https://jsonplaceholder.typicode.com/posts/1",json=data)
print(response.status_code)
print(response.json())
```

### 4. PATCH Request in Example

```python
data = {
    "body":"Trupti Currenlty PM of INDIA",
    "id":101
}


response = requests.patch("https://jsonplaceholder.typicode.com/posts/1",json=data)
print(response.status_code)
print(response.json())
```

### 5. DELETE Request in APIs

```python
response = requests.delete("https://jsonplaceholder.typicode.com/posts/1",json=data)
print(response.status_code)
print(response.json())
```
