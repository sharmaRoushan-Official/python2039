```python
# 10. Read a file and print each word on a separate line

# Path of the text file
fileName = r"C:\Users\User\Desktop\NursaryRhymes.txt"


# Open the file in read mode
with open(fileName, 'r') as file:

    # Read the entire file as a single string
    data = file.read()

    # Variable to store one word at a time
    word = ""


    # Read the file character by character
    for ch in data:

        # Check if the character is a space, newline, or tab
        if ch == " " or ch == "\n" or ch == "\t":

            # If the word is not empty, print it
            # (Better condition: if word != "")
            if word != " ":
                print(word)

                # Reset word for the next word
                word = ""

        else:
            # Add the current character to the word
            word += ch

# Note:
# To print the last word (if the file doesn't end with space/newline),
# add the following after the loop:
#
# if word != "":
#     print(word)
```

```python
## 12. Count the number of uppercase and lowercase letters in the file.

# Path of the text file
fileName = r"C:\Users\User\Desktop\NursaryRhymes.txt"

upperChar = 0
upperCharacter = []

lowerChar = 0
lowerCharacter = []
# Open the file in read mode
with open(fileName, 'r') as file:
    # print(file.read())
    for ch in file.read():
        # print(ch)
        if ch.isupper():
            # print(ch)
            upperCharacter.append(ch)
            upperChar+=1
        else:
            if ch == " " or ch == "\n" or ch == "\t":
                continue
            lowerChar +=1
            lowerCharacter.append(ch)

print("Upper chracters is:",upperCharacter,"=",upperChar)
print("Lower Chracters is:",lowerCharacter,"=",lowerChar)
```
