## 16. Decorator in Python

----

Q. What is Decorator?

A `decorator` is a function that modifies another function behavior without changing it's code.

📌 Define
> A Decorator adds extra feature to a function.


#### Why Decorator are Needed
❌ Without Decorators

If you want to:
* Log function call
* Check user authentication
* Measure execution time.

You would have to repeat the same code in every function.

✅️ With Decorator
*  Write logic once.
*  Reuse it many times.
*  Cleaner and maintable code.

### Real-Life Analogy

* Gift = Original Function
* Wrapper = Decorator


### Function are First-Class and Object


Basic Decorator Structure

```python
def decorator(func):  # func = parameter of function # a
    def wrapper():
        print("Before Function:")
        func() # a() say_hello()
        print("After Function")
    return wrapper
```

```python
def say_hello():
    print("Hello world")
x = say_hello()
```

## Applying A Decorator(Manual)

```python
decorator(x)
```

### using @ syntax [Most Important]

```python
@decorator
def greet():
    name = input("Enter your name: ")
    print("Congratulations! "+name)
greet()
```

```python
@decorator
def addTwo():
    a= 10
    b = 12
    sum1 = a+b
    print(f"Sum of two values:{sum1}")
addTwo()

@decorator
def greet():
    name = input("Enter your name: ")
    print("Congratulations! "+name)
greet()
```

Decorator with Arguments

```python
def decorator(func):
    def wrapper(name):
        print("Congratulations!!")
        func(name)
    return wrapper
        
```

```python
@decorator
def greet(name):
    print("Hello ",name)
greet("Gauri")
```

#### Decorator with *args and **kwargs

```python
def decoratorName(func):
    def wrapper(*args,**kwargs):
        print("Function Started")
        result = func(*args,**kwargs)
        print("Function Ended")
        return result
    return wrapper
```

```python
import time

def timer(func):
    def wrapper(*args,**kwargs):
        print("function started")
        start = time.time()
        result = func(*args,**kwargs)
        end = time.time()
        print("Exceution Time:",end-start)
        print("Function Finished")
        return result
    return wrapper
```

```python
# time.time()
```

```python
@timer
def demo():
    for i in range(1000000000):
        pass
demo()
```

### Decorator with Return Value

```python
def decorator(func):
    def wrapper():
        return func() + "!!!!!!"
    return wrapper
```

```python
@timer
@decorator
def message():
    return "Hello"
message()
```

#### Multiple Decorator

```python
def decor1(func):
    def wrapper():
        print("Decor 1")
        func()
    return wrapper

    
def decor2(func):
    def wrapper1():
        print("Decor 2")
        func()
    return wrapper1
```

```python
@decor2
@decor1
def show():
    print("Hello world")
show()
```
