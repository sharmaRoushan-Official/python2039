 7.Write a function factorial(n) that returns factorial of a number.


```python
def factorial(n):
    i=1
    fact=1
    for i in range(1,n+1):
        fact=fact*i
    print(f"factorial of {n} = {fact}")
```

```python
factorial(8)
```

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(8))
```

```python
def remove_duplicates(lst):
    new_list=[]
    for i in lst:
        if i not in new_list:
            new_list.append(i)
    return new_list

print(remove_duplicates([20,30,20,40,40,50]))
```

17.# Write a function unique_elements(lst) that returns only unique elements.


```python
def unique_elements(lst):
    new_list=[]
    for i in lst:
        if i not in new_list:
            new_list.append(i)
        else:
            new_list.remove(i)
            
    return new_list
    

print(unique_elements([20,30,20,40,50,40]))
```

18. Write a function second_largest(lst) that returns the second largest number.


```python
lst1 = [10, 20, 5, 30, 15]
def second_largest(lst):
    largest = lst[0]
    second = lst[0]

    for i in lst:
        if i > largest:
            second = largest
            largest = i
        elif i > second and i != largest:
            second = i

    return second

print(second_largest(lst1))
```

19.# Write a function prime_check(n) that returns True if number is prime.


```python
def prime_check(n):
    if n<2:
        return False
    for i in range(2,n):
        if n % i == 0:
            return False
    return True

print(prime_check(111))
```

20.# Write a function generate_primes(n) that generates all prime numbers up to n.


```python
def generate_primes(n):
    
    primes=[]
    
    for i in range(2, n+1):
        count=0
        for num in range(1,i+1):
            if i % num == 0:
                count+=1

        if count == 2:
            primes.append(num)
    return primes
    
print(generate_primes(50))
            
```

### 30. Write a function armstrong(n) that checks if number is Armstrong.


```python
def armstrong(n):
    total = 0
    for i in str(n):
        total = total + int(i) ** len(str(n))
        
    if total == n:
        print("Armstrong Number:",n)
    else:
        print("Not Armstrong Number:",n)

armstrong(133)
```

### 31  Write a function perfect_number(n) that checks if number is perfect.


```python
def perfect_number(n):
    sum1=0
    for i in range(1,n):
        if n%i==0:
            sum1=sum1+i
    if sum1==n:
        print(n," is perfect number")
    else:
        print(n,"is not a perfect number")
```

```python
perfect_number(6)
```

### 32.Write a function count_upper_lower(s) that counts uppercase and lowercase letters.


```python
def count_upper_lower(s):
    upper = 0
    lower = 0

    for i in s:
        if i >= 'A' and i <= 'Z':
            upper += 1
        elif i >= 'a' and i <= 'z':
            lower += 1

    return upper, lower

print(count_upper_lower("Hello World"))
```

### 33 Write a function sort_list(lst) without using built-in sort().


```python
def sort_list(lst):
    new_list=[]

    while lst:
        smallest = min(lst)
        new_list.append(smallest)
        lst.remove(smallest)
        
    return new_list
```

```python
sort_list([2,5,1,8,4,3])
```

### 34. Write a function binary_to_decimal(binary).


```python
def binary_to_decimal(num):
    decimal=0
    power=0

    for digits in reversed(num):
        decimal = decimal + int(digits)*(2**power)
        power+=1
    return decimal

print(binary_to_decimal("011010"))
```
