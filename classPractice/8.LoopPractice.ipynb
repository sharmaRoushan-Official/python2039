````python
### 5.Calculate the factorial of a number

# Factorial of n = n! = n × (n - 1) × (n - 2) × ... × 1
# 4! = 4*3*2*1 = 24
# 5! = 5*4*3*2*1 =120

num = int(input("Enter any number to get factorial: "))

fact = 1

steps = []
for i in range(1,num+1):
    fact = fact*i
    process = steps.append(fact)
print(fact)
print(steps)
````

````python
### 10. Print only vowels from a string input by user.

string1 = "The Quick Brown Fox Jump Over The Lazy Dog. 12233"
vowels = "AEIOUaeiou"

vowelsList = []
consonentList = []

vowCount = 0
consonentCount = 0
for i in range(len(string1)):
    if string1[i] in vowels:
        # print(string1[i])
        vowelsList.append(string1[i])
        vowCount+=1
    else:
        if string1[i] == " ":
            continue
        elif string1[i].isdigit():
            continue
        consonentList.append(string1[i])
        consonentCount+=1
print(f"Total Vowels in string is:",vowelsList,"=",vowCount)
print(f"Total Consonenet in string is:",consonentList,"=",consonentCount)
````

````python
### Using While Loop
string1 = "The Quick Brown Fox Jump Over The Lazy Dog. 12233"

i = 0
vowels = "AEIOUaeiou"

vowelsList = []
consonentList = []

vowCount = 0
consonentCount = 0

while i<len(string1)-1:
    i+=1
    if string1[i] in vowels:
        # print(string1[i])
        vowelsList.append(string1[i])
        vowCount+=1
    else:
        if string1[i] == " ":
            continue
        elif string1[i].isdigit():
            continue
        consonentList.append(string1[i])
        consonentCount+=1
print(f"Total Vowels in string is:",vowelsList,"=",vowCount)
print(f"Total Consonenet in string is:",consonentList,"=",consonentCount)    
````

````python
### 12. Take numbers as input untill the user enter 0, then print their sum of the value.


# infinite looop  using while loop and for loop

sumNum = 0
userNumList = []
for i in iter(int, 1): # iterable item  # 1:True
    userNum = int(input("Start Enter value: "))
    
    if userNum == 0:
        break
    userNumList.append(userNum)
    sumNum+=userNum
print(f"User Enter value is:",userNumList)
print(f"Total Sum of the value is:",sumNum)
````

````python
### using while loop
sumNum = 0
userNumList = []
while True:  # True it means infinine loop
    userNum = int(input("Start Enter value: "))
    
    if userNum == 0:
        break
    userNumList.append(userNum)
    sumNum+=userNum
print(f"User Enter value is:",userNumList)
print(f"Total Sum of the value is:",sumNum)
````

````python
num = 4

if num<=1:
    print("Not Prime")
else:
    prime = True

    for i in range(2,num):  # 2,4
        if num%i == 0:  
            prime = False
            break
    if prime:
        print("Prime Number",num)
    else:
        print("Not Prime",num)
````
