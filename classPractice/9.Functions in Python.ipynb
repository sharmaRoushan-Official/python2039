### 29.Find the GCD of two numbers using loop.


```python
num1= int(input("Enter a number 1 : "))
num2 = int(input("Enter a number 2 : "))
i=1
while i<=num1:
    if num1%i==0 and num2%i==0:
        gcd = i
    i=i+1
print(f"GCD of {num1} and {num2} is {gcd}")
```

```python
# LCM = (Number1*Number2)/GCD

num1= int(input("Enter a number 1 : "))
num2 = int(input("Enter a number 2 : "))

a = num1
b = num2

while b!=0:
    a,b = b, a%b
gcd = a
# print(gcd)
lcm = (num1*num2)//gcd
print("LCM:",lcm,"\nGCD:",gcd)
```

## 16. write a function remove_duplicates(lst) that removes duplicates from a list.



```python
lst1 = [10,12,13,10,14,12,15,13,16]  # [10,12,13,14,15,16]

uniqueNum = []
def uniqueNumber(lst):
    for i in lst:
        if i not in uniqueNum:
            uniqueNum.append(i)
    print("Unique Number is:",uniqueNum)
uniqueNumber(lst1)
```

### 22. Write a function count_occurrences(lst,item) that counts occurrences of an items.

```python
mylist = [10,20,30,20,40,20,50]  # static list


def count_occurrence(lst,item):
    count = 0
    for i in lst:
        # print(i)
        if i == item:
           count+=1
    print(count)
item = 20
count_occurrence(mylist,item)
```
