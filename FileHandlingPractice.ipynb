##### Write a table into file from 1-10 "table.txt"

```python
fileName = "table.txt"

with open(fileName,'w') as file:
    for i in range(1,11):  # table 1 to 10
        file.write(f"================Table of {i}================\n")
        for j in range(1,11): #1x1=1, 1x10=10
            file.write(f"{i}X{j}={i*j}")
            file.write("\n")
    file.write(f"End of Table from 1 to {j}")
```

```python
### Given sentence write in the file "sentence.txt" count total number of characters and words in the file.
### sentence = A qucik brown fox jump over the lazy dog.

### Total number of words = 9
### Total number of characters =  41
```

```python
fileName = "sentence.txt"
with open(fileName,"w") as file:
    sentence = "A qucik brown fox jump over the lazy dog."
    totalChar = 0
    words = sentence.split()
    wordCount = 0
    for word in words:
        wordCount+=1
        for char in word:
            # print(char)
            totalChar+=1
    # print("Total Words in your sentence: ",wordCount)
    # print("Total Characters in you sentence:", totalChar)
    file.write(f"Sentence:{sentence}\n")
    file.write(f"Total Words in your sentence: {wordCount}\n")
    file.write(f"Total Characters in Sentnce:{totalChar}\n")
```
