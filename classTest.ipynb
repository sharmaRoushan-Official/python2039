```python
### 1. write a function to count total numbers of 
### chracters, 
### total words,
### total Vowels, 
### total consonents, 
### total spaces 
### total UpperCase characters
### total LowerCase characters
### in the given sentence.
### sentence = "A Quick Brown Fox Jump Over the Lazy Dog"


def calculateNum():
    sentence = "A Quick Brown Fox Jump Over the Lazy Dog"
    i = 0
    totalChar = 0
    words = sentence.split()
    # print(words)
    vowChar = []
    countVow=0
    consChar = []
    consCount=0
    countSpace = 0
    charUpper = []
    countUpper = 0
    charLower = []
    countLower = 0
    while i<len(sentence)-1:
        # print(sentence[i])
        if sentence[i].isupper():
            charUpper.append(sentence[i])
            countUpper+=1
        else:
            if sentence[i].islower():
                charLower.append(sentence[i])
                countLower+=1
            
        i+=1
        totalChar = totalChar+1
        if sentence[i] in "AEIOUaeiou":
            vowChar.append(sentence[i])
            countVow+=1
        else:
            if sentence[i] == " ":
                countSpace+=1
                continue
            consChar.append(sentence[i])
            consCount+=1
    print("Total Characters:",totalChar)
    print("Total Words:",len(words))
    print("Total Vowel Characters:",vowChar)
    print("Total Vowel Numbers:",countVow)
    print("Total Consonent Characters:",consChar)
    print("Total Consonent Numbers:",consCount)
    print("Total Spaces:",countSpace)
    print("Total Upper Chracters:",charUpper)
    print("Total Upper Numbers",countUpper)
    print("Total Lower Characters:",charLower)
    print("Total Number of Lower:",countLower)
calculateNum()
```

```python
### 2.write a function Take infinite numbers from user and calculate their
### sum
### average
### total digits
### total even numbers
### sum of even number
### total odd numbers 
### sum of odd numbers


def calculateNumber():
    listNum = []
    sumNum = 0
    countNum = 0
    avgNum = 0  # 10
    evenNum = []
    evenSum = 0
    oddNum = []
    oddSum = 0
    while True:
        num = int(input("Enter Number: "))
        if num == 0:
            break
        listNum.append(num)
        sumNum+=num  # sumNum 10, 20
        countNum+=1  # 1
        avgNum = (sumNum/countNum)  # 30/2
        if num%2==0:
            evenNum.append(num)
            evenSum+=num
        else:
            oddNum.append(num)
            oddSum+=num
        
    print("List of Numbers:",listNum)
    print("Sum of Numbers:",sumNum)
    print("Count Numbers:",countNum)
    print("Average Value:",avgNum)
    print("Even Numbers:",evenNum)
    print("Even Sum:",evenSum)
    print("Odd Numbers:",oddNum)
    print("Odd Sum:",oddSum)
calculateNumber()
```
